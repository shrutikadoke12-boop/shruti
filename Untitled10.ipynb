```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
```

```python
(x_cifar_train, y_cifar_train),(x_cifar_test, y_cifar_test)= keras.datasets.cifar10.load_data()
```

```python
x_cifar_train = x_cifar_train.astype("float32") / 255.0
x_cifar_test = x_cifar_test.astype("float32") / 255.0
```

```python
y_cifar_train = y_cifar_train.squeeze();
y_cifar_test = y_cifar_test.squeeze()
```

```python
class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
print("CIFAR train:", x_cifar_train.shape, "test:", x_cifar_test.shape)
```

```python
plt.imshow(x_cifar_train[25]); 
plt.title(class_names[y_cifar_train[25]]);
plt.axis('off');
plt.show()
```

```python
cifar_model = keras.Sequential([
    keras.layers.Input(shape=(32,32,3)),
    keras.layers.Conv2D(32, 3, activation='relu', padding='same'),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(64, 3, activation='relu', padding='same'),
    keras.layers.MaxPooling2D(),
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(10, activation='softmax')
])
```

```python
cifar_model.compile(optimizer=keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
                    loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history_cifar = cifar_model.fit(x_cifar_train, y_cifar_train, batch_size=128, epochs=3,
                                validation_split=0.1, verbose=1)
cifar_loss, cifar_acc = cifar_model.evaluate(x_cifar_test, y_cifar_test, verbose=0)
```

```python
print(f"CIFAR-10 test loss: {cifar_loss:.4f}")
print(f"CIFAR-10 test accuracy: {cifar_acc:.4f}")
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,2,1);
plt.plot(history_cifar.history['loss'], label='train');
plt.plot(history_cifar.history['val_loss'], label='validation');
plt.title('CIFAR-10 loss'); plt.legend()
plt.subplot(1,2,2);
plt.plot(history_cifar.history['accuracy'], label='train');
plt.plot(history_cifar.history['val_accuracy'], label='validation');
plt.title('CIFAR-10 accuracy');
plt.legend()
plt.show()
```
