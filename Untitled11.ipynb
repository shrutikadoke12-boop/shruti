```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
print("Train:", x_train.shape, y_train.shape)
print("Test:", x_test.shape, y_test.shape)
```

```python
plt.imshow(x_train[7], cmap="gray");
plt.title(f"Label: {y_train[7]}");
plt.axis("off");
plt.show()
```

```python
mnist_model = keras.Sequential([
    keras.layers.Input(shape=(28, 28)),
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])
```

```python
mnist_model.compile(optimizer=keras.optimizers.SGD(learning_rate=0.01),
                    loss="sparse_categorical_crossentropy", metrics=["accuracy"])
EPOCHS = 3
history_mnist = mnist_model.fit(x_train, y_train, epochs=EPOCHS, batch_size=128,
                                validation_split=0.1, verbose=1)
loss_mnist, acc_mnist = mnist_model.evaluate(x_test, y_test, verbose=0)
print(f"MNIST test loss: {loss_mnist:.4f}")
print(f"MNIST test accuracy: {acc_mnist:.4f}")
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,2,1); plt.plot(history_mnist.history['loss'], label='train'); plt.plot(history_mnist.history['val_loss'], label='validation'); plt.title('MNIST loss'); plt.legend()
plt.subplot(1,2,2); plt.plot(history_mnist.history['accuracy'], label='train'); plt.plot(history_mnist.history['val_accuracy'], label='validation'); plt.title('MNIST accuracy'); plt.legend()
plt.show()
p = mnist_model.predict(x_test[:1], verbose=0).argmax(axis=1)[0]
print("First test image — actual:", y_test[0], "predicted:", p)
```
